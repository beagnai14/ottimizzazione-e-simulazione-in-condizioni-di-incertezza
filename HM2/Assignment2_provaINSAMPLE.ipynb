## Ottimizzazione e simulazione in condizioni di incertezza



Il modello Assembly to Order è un problema di programmazione stocastica a due stadi:    
- *primo stadio* (decisioni here-and-now): si decidono le quantità ottime di componenti base da produrre o acquistare prima di conoscere l'effettiva domanda dei prodotti finiti.   
- *Osservazione dell'incertezza*: La domanda si manifesta secondo una distribuzione di probabilità.
- *Secondo Stadio* (decisioni di recourse o wait-and-see): Una volta nota la domanda, decidi quanti prodotti finiti assemblare a partire dai componenti che si hanno a disposizione.

## Il modello:
Il problema è formulato come una massimizzazione del profitto atteso su un insieme di scenari $\mathcal{S}$. 

Insiemi:   
$\mathcal{I}$: insieme dei componenti.  
$\mathcal{J}$: insieme dei prodotti finiti.  
$\mathcal{M}$: insieme delle macchine.  
$\mathcal{S}$: insieme dei possibili esiti o scenari.  

Parametri:   
$C_i$: costo del componente $i$.    
$P_j$: prezzo di vendita del prodotto $j$.  
$L_m$: disponibilità temporale della macchina $m$.  
$T_{im}$: tempo necessario per produrre $i$ sulla macchina $m$.  
$G_{ij}$: fattore "gozinto", ovvero la quantità del componente $i$ necessaria per assemblare il prodotto $j$.  
$d_j^s$: domanda del prodotto $j$ nello scenario $s$.  
$\pi^s$: probabilità dello scenario $s$.  

Variabili:   
$x_i \ge 0$: quantità del componente $i$ prodotta (decisione here-and-now).   
$y_j^s \ge 0$: quantità del prodotto $j$ assemblata nello scenario $s$ (decisione wait-and-see).  


Formulazione:
$$\max \quad -\sum_{i \in \mathcal{I}} C_i x_i + \sum_{s \in \mathcal{S}} \pi^s \sum_{j \in \mathcal{J}} P_j y_j^s$$

Soggetto a:  
1) Vincolo di capacità produttiva: $\sum_{i \in \mathcal{I}} T_{im} x_i \le L_m \quad \forall m \in \mathcal{M}$.   
2) Vincolo di domanda massima: $y_j^s \le d_j^s \quad \forall j \in \mathcal{J}, \forall s \in \mathcal{S}$.    
3) Vincolo di disponibilità componenti (Distinta base): $\sum_{j \in \mathcal{J}} G_{ij} y_j^s \le x_i \quad \forall i \in \mathcal{I}, \forall s \in \mathcal{S}$.    
4) Non negatività: $y_j^s, x_i \ge 0$.

```python
# INSIEMI DEI PARAMETRI
components = [f'i{k}' for k in range(1, 16)]
products = ['j1', 'j2', 'j3']  # j1=Bike, j2=E-City, j3=E-MTB
machines = ['m1', 'm2']
scenarios = ['s1', 's2', 's3'] # Alta, Media, Bassa

# Probabilità degli scenari (pi^s)
pi = {'s1': 0.30, 's2': 0.50, 's3': 0.20}

# Costi dei componenti C[i]
C = {'i1': 200, 'i2': 400, 'i3': 325, 'i4': 575, 'i5': 100, 'i6': 200, 'i7': 40, 
     'i8': 25, 'i9': 70, 'i10': 50, 'i11': 175, 'i12': 325, 'i13': 80, 'i14': 30, 'i15': 20}

# Prezzi di vendita P[j]
P = {'j1': 999, 'j2': 1999, 'j3': 2999}

# Capacità macchine L[m] in minuti
L = {'m1': 15000, 'm2': 15000}

# Tempi di produzione T[i][m]
T = {
    # Componente: { m1, m2}
    'i1':  {'m1': 20,  'm2': 40},
    'i2':  {'m1': 25,  'm2': 55},
    'i3':  {'m1': 0,  'm2': 25},
    'i4':  {'m1': 0,  'm2': 35},
    'i5':  {'m1': 0,  'm2': 15},
    'i6':  {'m1': 20, 'm2': 0},
    'i7':  {'m1': 5,  'm2': 0},
    'i8':  {'m1': 10, 'm2': 0},
    'i9':  {'m1': 15, 'm2': 0},
    'i10': {'m1': 15, 'm2': 0},
    'i11': {'m1': 20, 'm2': 0},
    'i12': {'m1': 60, 'm2': 0},
    'i13': {'m1': 0,  'm2': 15},
    'i14': {'m1': 5,  'm2': 5},
    'i15': {'m1': 8,  'm2': 0}
}

#Quantià di i in j G[i][j]
G = {  
    'i1':  {'j1': 0, 'j2': 1, 'j3': 0},
    'i2':  {'j1': 0, 'j2': 0, 'j3': 1},
    'i3':  {'j1': 0, 'j2': 1, 'j3': 0},
    'i4':  {'j1': 0, 'j2': 0, 'j3': 1},
    'i5':  {'j1': 0, 'j2': 1, 'j3': 1},
    'i6':  {'j1': 2, 'j2': 2, 'j3': 2},
    'i7':  {'j1': 1, 'j2': 1, 'j3': 1},
    'i8':  {'j1': 1, 'j2': 1, 'j3': 0},
    'i9':  {'j1': 0, 'j2': 0, 'j3': 1},
    'i10': {'j1': 2, 'j2': 2, 'j3': 0},
    'i11': {'j1': 0, 'j2': 0, 'j3': 2},
    'i12': {'j1': 1, 'j2': 1, 'j3': 1},
    'i13': {'j1': 0, 'j2': 1, 'j3': 1},
    'i14': {'j1': 2, 'j2': 2, 'j3': 2},
    'i15': {'j1': 1, 'j2': 1, 'j3': 0}
}
```

```python
import numpy as np


def target_to_lognormal(target_mean, target_std):
    """
    Converte la media e la deviazione standard desiderate (spazio reale) 
    nei parametri mu e sigma richiesti per generare la distribuzione Lognormale.
    """
    variance = target_std ** 2
    mu_log = np.log(target_mean ** 2 / np.sqrt(variance + target_mean ** 2))
    sigma_log = np.sqrt(np.log(1 + variance / target_mean ** 2))
    return mu_log, sigma_log


def genera_domande_per_scenario(sample_size):
    # s1: ALTA DOMANDA (Lognormale)
    mu_s1_j1, sig_s1_j1 = target_to_lognormal(300, 45)
    d_s1_j1 = np.round(np.random.lognormal(mu_s1_j1, sig_s1_j1, sample_size)).astype(int)
    mu_s1_j2, sig_s1_j2 = target_to_lognormal(200, 30)
    d_s1_j2 = np.round(np.random.lognormal(mu_s1_j2, sig_s1_j2, sample_size)).astype(int)
    mu_s1_j3, sig_s1_j3 = target_to_lognormal(150, 25)
    d_s1_j3 = np.round(np.random.lognormal(mu_s1_j3, sig_s1_j3, sample_size)).astype(int)
    
    # s2: MEDIA DOMANDA (Normale Troncata a 0)
    d_s2_j1 = np.round(np.maximum(0, np.random.normal(225, 20, sample_size))).astype(int)
    d_s2_j2 = np.round(np.maximum(0, np.random.normal(140, 15, sample_size))).astype(int)
    d_s2_j3 = np.round(np.maximum(0, np.random.normal(100, 10, sample_size))).astype(int)

    # s3: BASSA DOMANDA (Lognormale)
    mu_s3_j1, sig_s3_j1 = target_to_lognormal(150, 30)
    d_s3_j1 = np.round(np.random.lognormal(mu_s3_j1, sig_s3_j1, sample_size)).astype(int)
    mu_s3_j2, sig_s3_j2 = target_to_lognormal(80, 20)
    d_s3_j2 = np.round(np.random.lognormal(mu_s3_j2, sig_s3_j2, sample_size)).astype(int)
    mu_s3_j3, sig_s3_j3 = target_to_lognormal(50, 15)
    d_s3_j3 = np.round(np.random.lognormal(mu_s3_j3, sig_s3_j3, sample_size)).astype(int)


        
    # TRASFORMAZIONE IN SCALARI: Integrazione Monte Carlo tramite media campionaria intera
    d = {
        'j1': {
            's1': int(np.round(np.mean(d_s1_j1))),
            's2': int(np.round(np.mean(d_s2_j1))),
            's3': int(np.round(np.mean(d_s3_j1)))
        },
        'j2': {
            's1': int(np.round(np.mean(d_s1_j2))),
            's2': int(np.round(np.mean(d_s2_j2))),
            's3': int(np.round(np.mean(d_s3_j2)))
        },
        'j3': {
            's1': int(np.round(np.mean(d_s1_j3))),
            's2': int(np.round(np.mean(d_s2_j3))),
            's3': int(np.round(np.mean(d_s3_j3)))
        }
    }
        
    return d

```

```python
# INIZIALIZZAZIONE DEL MODELLO 
import gurobipy as gp 
from gurobipy import GRB


def solve_bike_ato(sample_size):

    # creo il modello 
    model = gp.Model("Bike_ATO")

    model.Params.OutputFlag = 0


    sample_indices = range(sample_size)
    
    # creo le variabili >=0
    x = model.addVars(components, vtype=GRB.INTEGER, lb=0, name="x")
    y = model.addVars(products, scenarios, vtype=GRB.INTEGER, lb=0, name="y")


    # FUNZIONE OBIETTIVO

    # Costo certo di produzione dei componenti (Stadio 1)
    costo_produzione = gp.quicksum(
        C[i] * x[i] 
        for i in components
    )

    # Ricavo atteso ponderato per le probabilità (Stadio 2)
    ricavo_atteso = gp.quicksum(
        pi[s] * P[j] * y[j, s] 
        for s in scenarios 
        for j in products
    )

    # Massimizzazione del Profitto Atteso
    model.setObjective(ricavo_atteso - costo_produzione, GRB.MAXIMIZE)


    # VINCOLI 

    # 1) Vincolo di capacità macchine (Stadio 1)
    for m in machines:
        model.addConstr(
            gp.quicksum(T[i][m] * x[i] for i in components) <= L[m],
            name=f"Capacita_{m}"
        )


    # Genero una singola domanda per ogni coppia prodotto-scenario
    # d[j][s] è uno scalare
    d = genera_domande_per_scenario(sample_size)


    # 2) Vincolo di domanda massima (Stadio 2)
    for j in products:
        for s in scenarios:
            model.addConstr(
                y[j, s] <= d[j][s],
                name=f"Domanda_{j}_{s}"
            )


    # 3) Vincolo di disponibilità componenti 
    for i in components:
        for s in scenarios:
            model.addConstr(
                gp.quicksum(G[i][j] * y[j, s] for j in products) <= x[i],
                name=f"Disponibilita_{i}_{s}"
            )


    # Risolvo il modello
    model.optimize()


    # Salvo le soluzioni delle variabili di primo stadio
    x_sol = {
        i: x[i].X 
        for i in components
    }


    # Salvo le soluzioni delle variabili di secondo stadio
    y_sol = {
        (j, s): y[j, s].X 
        for j in products 
        for s in scenarios
    }


    # Restituisco risultati principali
    return {
        "objective": model.ObjVal,
        "x": x_sol,
        "y": y_sol,
        "d": d,
        "model": model
    }
```

```python
# STUDIO DELLA STABILITÀ IN-SAMPLE

# Parametri dello studio
sample_size = 30              # cardinalità iniziale degli insiemi S e S'
sample_size_max = 100        # cardinalità massima da testare
N = 100                       # numero di confronti indipendenti tra S e S'
alpha = 0.05                 # livello di significatività
z = 1.96                     # valore critico per alpha = 0.05

epsilon = 10 

# Lista in cui salvo i risultati
risultati_stabilita = []

# Condizione iniziale
stabile = False


# Aumento sample_size di 1 in 1 finché non raggiungo la stabilità
while (not stabile) and (sample_size <= sample_size_max):

    # Lista dei valori phi_i
    phi_values = []


    # Per ogni sample_size faccio N confronti indipendenti
    for i in range(N):

        # Risolvo il modello su un primo insieme casuale di scenari S
        res_1 = solve_bike_ato(sample_size)

        # Risolvo il modello su un secondo insieme casuale di scenari S'
        res_2 = solve_bike_ato(sample_size)


        # Calcolo phi_i come differenza tra i due valori ottimi
        # Nel nostro caso il modello massimizza il profitto atteso
        phi_i = res_1["objective"] - res_2["objective"]

        # Salvo phi_i
        phi_values.append(phi_i)


    # Calcolo la media campionaria delle phi_i
    phi_bar = np.mean(phi_values)

    # Calcolo la deviazione standard campionaria delle phi_i
    sigma_phi = np.std(phi_values, ddof=1)

    # Calcolo il semi-intervallo di confidenza
    half_width = z * sigma_phi / np.sqrt(N)

    # Estremo inferiore dell'intervallo di confidenza
    lower_bound = phi_bar - half_width

    # Estremo superiore dell'intervallo di confidenza
    upper_bound = phi_bar + half_width


    # Condizione di arresto: lo zero appartiene all'intervallo di confidenza
    stabile = (lower_bound <= 0 <= upper_bound)


    # Salvo i risultati di questo sample_size
    risultati_stabilita.append({
        "sample_size": sample_size,
        "N": N,
        "phi_bar": phi_bar,
        "sigma_phi": sigma_phi,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "stabile": stabile
    })


    # Stampo un riepilogo
    print(
        f"sample_size = {sample_size} | "
        f"N = {N} | "
        f"phi_bar = {phi_bar:.2f} | "
        f"IC = [{lower_bound:.2f}, {upper_bound:.2f}] | "
        f"stabile = {stabile}"
    )


    # Se non ho ancora raggiunto la stabilità, aumento sample_size di 1
    if not stabile:
        sample_size += 1
```

```python
# --- NUOVI PARAMETRI ---
epsilon = 600.0  # Soglia di tolleranza (es. 50 euro di differenza nel profitto)
# -----------------------

sample_size = 30
stabile = False

while (not stabile) and (sample_size <= sample_size_max):
    
    phi_values = []
    
    for i in range(N):
        res_1 = solve_bike_ato(sample_size)
        res_2 = solve_bike_ato(sample_size)
        
        # Calcoliamo la differenza assoluta
        diff = abs(res_1["objective"] - res_2["objective"])
        phi_values.append(diff)
    
    # Calcoliamo la media delle differenze assolute
    mean_diff = np.mean(phi_values)
    
    # CONDIZIONE DI ARRESTO: la media delle differenze è < epsilon
    stabile = (mean_diff <= epsilon)
    
    print(f"sample_size = {sample_size} | mean_diff = {mean_diff:.2f} | stabile = {stabile}")
    
    if not stabile:
        sample_size += 1
```
